# Objectives

# Objective 1

```python
#Creating a bar plot using Seaborn to visualize this relationship
df["DateTime"] = pd.to_datetime(df["DateTime"])
df = df.dropna(subset=["DateTime"])
df["Hour"] = df["DateTime"].dt.hour
df["Crash"] = 1  # Adding a column to count crashes
hourly_df = df.groupby("Hour")["Crash"].count().reset_index()

plt.figure(figsize=(12, 6))
sns.barplot(x=hourly_df["Hour"], y=hourly_df["Crash"], alpha=0.8, legend=True)
plt.title("Crash Frequency by Hour of Day (Seaborn)")
plt.xlabel("Hour of Day (0 = Midnight)")
plt.ylabel("Number of Crashes")
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.xticks(range(0, 24))
plt.tight_layout()
plt.show()
```

# Objective 2


```python
# Bar Plot: Top 10 Crash-Prone Streets
top_streets = df['StreetName'].value_counts().nlargest(10).reset_index()
top_streets.columns = ['Street Name', 'Number of Crashes']

plt.figure(figsize=(10, 6))
sns.barplot(data=top_streets, x='Number of Crashes', y='Street Name', palette='crest')
plt.title('Top 10 Crash-Prone Streets')
plt.xlabel('Number of Crashes')
plt.ylabel('Street Name')
plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Loading data
df = pd.read_csv("1.08_Crash_Data_Report_(detail).csv")

# Scattering Plot without dropping rows
plt.figure(figsize=(10, 5))
plt.scatter(df["Age_Drv1"], df["Distance"], color="darkgrey")
plt.xlabel("Driver Age")
plt.ylabel("Crash Distance (ft)")
plt.title("Crash Distance vs Driver Age")
plt.xscale("log")
plt.yscale("log")
plt.grid()
plt.show()
```

# Objective 4

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Converting to datetime (if not already done)
df['DateTime'] = pd.to_datetime(df['DateTime'], errors='coerce')

# Extracting month
df['Month'] = df['DateTime'].dt.month

# Counting crashes per month
monthly_crashes = df['Month'].value_counts().sort_index()

# Plotting
plt.figure(figsize=(10, 6))
sns.lineplot(x=monthly_crashes.index, y=monthly_crashes.values, marker='o', linewidth=2.5)
plt.title('Monthly Trend of Crashes', fontsize=16)
plt.xlabel('Month (1 = Jan, 12 = Dec)')
plt.ylabel('Number of Crashes')
plt.xticks(range(1, 13))
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

# Objective 5


```python
# Heatmap: Crash Severity by Hour of Day
plt.figure(figsize=(12, 6))
ct = pd.crosstab(df['DateTime'].dt.hour, df['Injuryseverity'])
sns.heatmap(ct, cmap='YlOrBr', annot=True, fmt='d')
plt.title('Crash Severity by Hour of Day')
plt.xlabel('Injury Severity')
plt.ylabel('Hour of Day')
plt.tight_layout()
plt.show()
```
